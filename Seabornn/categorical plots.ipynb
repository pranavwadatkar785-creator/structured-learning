```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
irir = sns.load_dataset("iris")
```

```python

sns.scatterplot(data=tips,x="total_bill",y="tip",hue="sex")
```

```python
#categorical scatter plot
sns.stripplot(data=tips,x="day",y="total_bill",hue="day") # try ,jitter=False
```

```python
#figure lvl
#1) stripplot
sns.catplot(kind="strip",data=tips,x="day",y="total_bill",hue="day")
```

```python
#2) swarmplot 
sns.catplot(kind="swarm",data=tips,x="day",y="total_bill",hue="sex")
```

```python
#categorical distribution plots
#1) boxplot
# helps determine outliers in the data
# also helps determine if data symmetrical or not

sns.boxplot(data=tips,x="day",y="total_bill",hue="sex")
```

```python
sns.boxplot(data=tips,y="total_bill")
```

```python
#2) voilenplot(box + kde)
sns.violinplot(data=tips,x="day",y="total_bill",hue="day")
```

```python
sns.catplot(kind="violin",data=tips,x="day",y="total_bill",hue="sex",split=True)
```

```python
#categorical estimate plot
#1) barplot
# plt.figure(figsize=(50,10))
plt.xticks(rotation=45)
sns.barplot(data=tips,x="sex",y="total_bill")
```

```python
import numpy as np
sns.catplot(kind="bar",data=tips,x="sex",y="total_bill",hue="smoker",estimator=np.min)
```

```python
#pointplot
sns.catplot(kind="point",data=tips,x="sex",y="total_bill",hue="smoker",estimator=np.min)
```

```python
sns.catplot(kind="point",data=tips,x="sex",y="total_bill",errorbar=None)
```

```python
#countplot
sns.catplot(kind="count",data=tips,x="sex",hue="day",col="smoker")
```
