```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly as px
import pandas as pd
```

```python
gap = px.data.gapminder("gap")
gap["year"] = pd.to_datetime(gap["year"]).dt.year
gap
```

```python
temp_df = gap.pivot(index="country",columns="year",values="lifeExp")
```

```python
#heatmap
plt.figure(figsize=(15,10))
sns.heatmap(data=temp_df,linewidths=0.5,cmap="viridis")
```

```python
temp_df = gap[gap["continent"]=="Asia"]
temp_df = temp_df.pivot(index="country",columns="year",values="lifeExp")

plt.figure(figsize=(15,10))
sns.heatmap(data=temp_df,annot=True,linewidths=0.5,cmap="viridis")
```
