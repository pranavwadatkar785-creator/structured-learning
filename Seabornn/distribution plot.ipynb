```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly as px

#distribution plot for 1)histplot 2)kdeplot 3)rugplot
#rugplot cannot used using displot
```

```python
#histplot,kdeplot,rugplot obviously axis lvl
#displot figure lvl func
tips = sns.load_dataset("tips")
sns.histplot(data=tips,x="total_bill",bins=2)
```

```python
sns.displot(kind="hist",data=tips,x="total_bill",bins=20)
```

```python
sns.displot(kind="hist",data=tips,x="tip",hue="sex")
```

```python
sns.displot(kind="hist",data=tips,x="tip",hue="sex",element="step")
```

```python
titanic = sns.load_dataset("titanic")
titanic
```

```python
sns.displot(kind="hist",data=titanic,x="age",hue="alive",col="sex",element="step")
#faceting will not work on axis lvl functions
```

```python
#kdeplot

sns.kdeplot(data=tips,x="total_bill")
```

```python
sns.displot(kind="kde",data=titanic,x="age",hue="alive",fill=True,col="sex",cut=0)
```

```python
#rug plot
g = sns.displot(kind="kde",data=titanic,x="age",col="sex",fill=True,cut=0)

for i in g.axes.flat:
    sns.rugplot(data=titanic,x="age",ax=i)
```

```python
sns.histplot(data=titanic,x="age",y="alive",color="red")
```

```python

sns.displot(kind="kde",data=tips,x="total_bill",y="tip")
```
