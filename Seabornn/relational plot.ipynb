```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly as px

#relational plot for scatter and line plot
```

```python
tips = sns.load_dataset("tips")
tips
```

```python
#scatter plot (is an axis level function)
sns.scatterplot(data=tips,x="total_bill",y="tip")

#relplot (figure level function)
sns.relplot(data=tips,x="total_bill",y="tip", kind="scatter")
```

```python
#scatter plot (is an axis level function)
sns.scatterplot(data=tips,x="total_bill",y="tip",hue="sex",style="time",size="size")

#relplot (figure level function)
sns.relplot(data=tips,x="total_bill",y="tip", kind="scatter",hue="sex",style="time",size="size")
```

```python
#line plot 
gap = px.data.gapminder("gap")
gap_ind = gap[gap["country"]=="India"]
gap_ind
```

```python
sns.lineplot(data=gap_ind,x="year",y="lifeExp",) #axis lvl func

sns.relplot(data=gap_ind,x="year",y="lifeExp",kind="line") #figure lvl func
```

```python
gap_compare = gap[gap["country"].isin(["India","China","United States","Italy"])]

sns.relplot(kind="line",data=gap_compare,x="year",y="lifeExp",hue="country",style="continent")
```

```python
#facet plot will only work with figure lvl functions
sns.relplot(data=tips,kind="scatter",x="total_bill",y="tip",col="sex",row="smoker")
```

```python
sns.relplot(data=gap,x="lifeExp",y="gdpPercap",kind="scatter",col="year",col_wrap=3)
```
