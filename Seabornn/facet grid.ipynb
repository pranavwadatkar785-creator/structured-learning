```python
import seaborn as sns
import pandas as pd
import numpy as pd
import matplotlib.pyplot as plt

#********facet plots only work with axis lvl functions************
```

```python
tips = sns.load_dataset("tips")
```

```python
sns.catplot(data=tips,x="sex",y="total_bill",hue="sex",kind="violin",col="day",row="smoker")
```

```python
#using facet grid

g = sns.FacetGrid(data=tips,hue="sex",col="day",row="smoker")
g.map(sns.boxplot,"sex","total_bill")
```
