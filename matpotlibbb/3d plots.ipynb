```python
#3d surface plots

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx, yy = np.meshgrid(x,y)

z = xx**2 + yy**2
```

```python
fig =plt.figure(figsize=(12,8))

ax = plt.subplot(projection="3d")
ax.plot_surface(xx,yy,z,cmap="viridis")
fig.colorbar(ax.plot_surface(xx,yy,z,cmap="viridis"))
```

```python
z = np.sin(xx) + np.log(yy)

ax = plt.subplot(projection="3d")
ax.plot_surface(xx,yy,z,cmap="viridis")
fig.colorbar(ax.plot_surface(xx,yy,z,cmap="viridis"))
```

```python
#contor graph
z = xx**2 + yy**2

fig =plt.figure(figsize=(12,8))

ax = plt.subplot()
ax.contour(xx,yy,z,cmap="viridis")
fig.colorbar(ax.contour(xx,yy,z,cmap="viridis"))
```

```python
#contorf graph
z = xx**2 + yy**2

fig =plt.figure(figsize=(12,8))

ax = plt.subplot()
ax.contourf(xx,yy,z,cmap="viridis")
fig.colorbar(ax.contourf(xx,yy,z,cmap="viridis"))
```

```python
#contorf graph
z = np.sin(xx) + np.cos(yy)

fig =plt.figure(figsize=(12,8))

ax = plt.subplot()
ax.contourf(xx,yy,z,cmap="viridis")
fig.colorbar(ax.contourf(xx,yy,z,cmap="viridis"))
```
