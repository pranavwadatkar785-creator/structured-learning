```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
price = [45000,54000,58000,49000,47500,42000]
year = [2021,2022,2023,2024,2025,2026]

plt.bar(year,price,color="grey")
plt.xlabel("Year")
plt.ylabel("Price")
```

```python
df = pd.read_csv(r"C:\Users\Pranav Wadatkar\Desktop\Python\structured-learning\data\matplot\batsman_season_record.csv")

#for multiple bar charts
plt.bar(np.arange(df.shape[0]) - 0.2,df["2015"],width=0.2,label="2015")
plt.bar(np.arange(df.shape[0]),df["2016"],width=0.2,label="2016")
plt.bar(np.arange(df.shape[0]) + 0.2,df["2017"],width=0.2,label="2017")

plt.xticks(np.arange(df.shape[0]),df["batsman"])
plt.legend()
plt.show()
```

```python
price = [45000,54000,58000,49000,47500,42000]
year = ["2021111111111111111111111111111111111","2022","2023","2024","2025","2026"]

plt.bar(year,price,color="grey")
plt.xlabel("Year")
plt.ylabel("Price")
plt.xticks(rotation=90)
```

```python
#stacked bar charts
plt.bar(df["batsman"],df["2015"], label="2015")
plt.bar(df["batsman"],df["2016"],bottom=df["2015"], label="2016")
plt.bar(df["batsman"],df["2017"],bottom=(df["2015"] + df["2016"]), label="2017")

plt.legend()
```
