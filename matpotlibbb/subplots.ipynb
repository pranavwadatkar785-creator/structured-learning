```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
batter = pd.read_csv(r"C:\Users\Pranav Wadatkar\Desktop\Python\structured-learning\data\matplot\batter.csv")
batter.head()
```

```python
fig, ax = plt.subplots(figsize=(7,5))

ax.scatter(batter["avg"],batter["strike_rate"])
ax.set_title("Title")
ax.set_xlabel("Avg")
ax.set_ylabel("Strike Rate")
```

```python
fig, ax = plt.subplots(nrows=1,ncols=2,figsize=(15,7))
ax[0].scatter(batter["avg"],batter["strike_rate"])
ax[0].set_title("Title")
ax[0].set_xlabel("Avg")
ax[0].set_ylabel("Strike Rate")

ax[1].scatter(batter["avg"],batter["runs"])
ax[1].set_title("Title")
ax[1].set_xlabel("Avg")
ax[1].set_ylabel("Runs")
```

```python
#print(batter[batter["runs"] < 1500].to_string())

fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(15,7))
ax[0,0].scatter(batter["avg"],batter["strike_rate"])
ax[0,0].set_title("Title")
ax[0,0].set_ylabel("Strike Rate")

ax[1,0].scatter(batter["avg"],batter["runs"],color="red")
ax[1,0].set_title("Title")
ax[1,0].set_xlabel("Avg")
ax[1,0].set_ylabel("Runs")

ax[0,1].hist(batter["strike_rate"])
ax[1,1].hist(batter["runs"])
```
