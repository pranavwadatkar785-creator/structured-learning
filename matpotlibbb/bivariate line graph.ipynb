```python
batsman = pd.read_csv(r"C:\Users\Pranav Wadatkar\Desktop\Python\structured-learning\data\matplot\sharma-kohli.csv")

plt.plot(batsman["index"],batsman["V Kohli"],color="red")
plt.plot(batsman["index"],batsman["RG Sharma"],color="blue",linestyle="dashed",linewidth="3")

#title
plt.title("Rohit Sharma Vs Virat Kholi Runs Comparion")

#labels
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
#marker
plt.plot(batsman["index"],batsman["V Kohli"],color="red",marker="D",markersize="3")
plt.plot(batsman["index"],batsman["RG Sharma"],color="blue",linestyle="dashed",linewidth="3",marker="o")

#title
plt.title("Rohit Sharma Vs Virat Kholi Runs Comparion")

#labels
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
#label
plt.plot(batsman["index"],batsman["V Kohli"],color="red",marker="D",markersize="3",label="V Kholi")
plt.plot(batsman["index"],batsman["RG Sharma"],color="blue",linestyle="dashed",linewidth="3",marker="o",label="RG Sharma")
plt.legend()

#title
plt.title("Rohit Sharma Vs Virat Kholi Runs Comparion")

#labels
plt.xlabel("Season")
plt.ylabel("Runs Scored")

#grid
plt.grid()

plt.show()
```
